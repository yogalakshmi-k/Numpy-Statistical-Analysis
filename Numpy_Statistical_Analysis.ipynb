```python

import numpy as np

marks = np.array([45, 50, 55, 60, 65, 70, 75, 80, 85, 100])

print("Student Marks:", marks)
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Standard Deviation:", np.std(marks))
print("Minimum Marks:", np.min(marks))
print("Maximum Marks:", np.max(marks))

print("\nObservations:")
print("1. The average mark is 68.5.")
print("2. The median is 67.5, which is close to the mean.")
print("3. The mark 100 is unusually high compared with the lower marks.")
```
